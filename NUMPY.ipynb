**TRIGONOMETRIC CURVES**



```python
x = np.arange(1,11)
y = np.arange(10, 110, 10)

plt.figure(figsize = (5,3))
plt.plot(x,y, 'o--')
plt.show()

```

**SIN CURVE**

```python
x_sin = np.arange(0,2*np.pi, 0.1)
y_sin = np.sin(x_sin)
print(y_sin)
```

```python
plt.figure(figsize = (5,3))
plt.plot(x_sin, y_sin, 'y-')
plt.title("Sin Wave")
plt.show()
```

**COS CURVE**

```python
x_cos = np.arange(0,2*np.pi, 0.1)
y_cos = np.cos(x_cos)
print(y_cos)
```

```python
plt.figure(figsize = (5,3))
plt.plot(x_cos, y_cos, 'y-')
plt.title("Cos Wave")
plt.show()
```

**TAN CURVE**

```python
x_tan = np.arange(0,2*np.pi, 0.1)
y_tan =np.tan(x_tan)
print(y_tan)
```

```python
plt.figure(figsize = (5,3))
plt.plot(x_tan, y_tan, 'y-')
plt.title("Tan Wave")
plt.show()
```

**COT CURVE**

```python
x_cot = np.arange(0,2*np.pi, 0.1)
y_cot =1/np.tan(x_cot)
print(y_cot)
```

```python
plt.figure(figsize = (5,3))
plt.plot(x_cot, y_cot, 'y-')
plt.title("Cot Wave")
plt.show()
```

**SEC CURVE**

```python
x_sec = np.arange(0,2*np.pi, 0.1)
y_sec = 1/np.cos(x_sec)
print(y_sec)
```

```python
plt.figure(figsize = (5,3))
plt.plot(x_sec, y_sec, 'y-')
plt.title("Sec Wave")
plt.show()
```

**COSEC**

```python
x_cosec = np.arange(0,2*np.pi, 0.1)
y_cosec = 1/np.sin(x_cosec)
print(y_cosec)
```

```python
plt.figure(figsize = (5,3))
plt.plot(x_cosec, y_cosec, 'y-')
plt.title("Cosec Wave")
plt.show()
```

**6 CURVES TOGETHER**

```python
plt.figure(figsize = (6,6))
plt.subplot(3,2,1)
plt.plot(x_sin, y_sin, 'r-')
plt.title("Sin Wave")

plt.subplot(3,2,2)
plt.plot(x_cos, y_cos, 'b-')
plt.title("Cos Wave")

plt.subplot(3,2,3)
plt.plot(x_tan, y_tan, 'g-')
plt.title("Tan Wave")

plt.subplot(3,2,4)
plt.plot(x_cot, y_cot, 'y-')
plt.title("Cot Wave")

plt.subplot(3,2,5)
plt.plot(x_sec, y_sec, 'y-')
plt.title("Sec Wave")

plt.subplot(3,2,6)
plt.plot(x_cosec, y_cosec, 'y-')
plt.title("Cosec Wave")

plt.show()
```

EMPLOYEE SALARY ANALYSIS

```python
import matplotlib.pyplot as plt

salary = [25000, 30000, 35000, 40000, 45000, 50000]
experience = [1, 2, 3, 4, 5, 6]

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(salary)
plt.title("Salary Histogram")

plt.subplot(1,2,2)
plt.scatter(experience, salary)
plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.tight_layout()
plt.show()
```

IPL STATISTICS DASHBOARD


```python
import matplotlib.pyplot as plt

teams = ['MI', 'CSK', 'RCB', 'KKR']
runs = [180, 200, 170, 190]
wins = [5, 4, 3, 6]

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.bar(teams, runs)
plt.title("Runs")

plt.subplot(1,3,2)
plt.pie(wins, labels=teams, autopct='%1.1f%%')
plt.title("Team Wins")

plt.subplot(1,3,3)
matches = [1,2,3,4]
scores = [180, 200, 170, 190]
plt.plot(matches, scores, marker='o')
plt.title("Match Scores")

plt.tight_layout()
plt.show()
```
